### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python


# Construct metric for fatal/serious injuries

# Cleaning assumption:
# Convert fatal injury counts to numeric values.
# Any values that cannot be converted become NaN rather than
# being incorrectly treated as zero.

df_filtered['Total.Fatal.Injuries'] = pd.to_numeric(
    df_filtered['Total.Fatal.Injuries'],
    errors='coerce'
)

# Inspect available passenger and injury columns before
# estimating the total number of passengers and injury rates.
print([
    col for col in df_filtered.columns
    if 'Passenger' in col or 'Injury' in col
])
```

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python

# Aircraft.Damage
# Cleaning assumption: treat missing or unrecognized damage values
# as unknown rather than assuming the aircraft was not destroyed.

df_filtered['Aircraft.damage'] = (
    df_filtered['Aircraft.damage']
    .astype('string')
    .str.strip()
    .replace('', pd.NA)
)

# Derived column: True if the aircraft was destroyed,
# False if it had another known damage status, and <NA> if unknown.
df_filtered['Aircraft.Destroyed'] = (
    df_filtered['Aircraft.damage']
    .str.casefold()
    .map({'destroyed': True})
    .fillna(
        df_filtered['Aircraft.damage'].notna()
        & ~df_filtered['Aircraft.damage'].str.casefold().eq('destroyed')
    )
    .astype('boolean')
)

print(df_filtered[['Aircraft.damage', 'Aircraft.Destroyed']].head(10))
print("\nMissing damage values:", df_filtered['Aircraft.damage'].isna().sum())
```

### Investigate the *Make* column
- Identify cleaning tasks here
- List cleaning tasks clearly in markdown
- Execute the cleaning tasks
- For your analysis, keep Makes with a reasonable number (you can put the threshold at 50 though lower could work as well)


### Cleaning assumptions for the Make column

- Remove leading and trailing spaces from aircraft makes.
- Treat empty strings as missing values.
- Exclude records with missing aircraft makes because they cannot be reliably identified.
- Group make names without distinguishing uppercase and lowercase letters when counting records.
- Retain only aircraft makes with at least 50 records, as suggested in the assignment.
- This threshold reduces the influence of makes with very few observations on the analysis.

```python


# Clean and standardize aircraft make names
df_filtered['Make'] = (
    df_filtered['Make']
    .astype('string')
    .str.strip()
    .str.upper()
    .replace('', pd.NA)
)

# Remove records with missing aircraft makes
df_filtered = df_filtered.dropna(subset=['Make']).copy()

# Count each standardized make
make_counts = df_filtered['Make'].value_counts()

# Retain makes with at least 50 records
valid_makes = make_counts[make_counts >= 50].index
df_filtered = df_filtered[
    df_filtered['Make'].isin(valid_makes)
].copy()

# Display cleaning results
print("Number of records after cleaning:", len(df_filtered))
print("\nAircraft makes retained:")
print(df_filtered['Make'].value_counts().head(20))
print("\nNumber of distinct makes retained:",
      df_filtered['Make'].nunique())
```

### Inspect Model column
- Get rid of any NaNs.
- Inspect the column and counts for each model/make. Are model labels unique to each make?
- If not, create a derived column that is a unique identifier for a given plane type.

```python

# Inspect and clean the Model column

# Remove rows where the aircraft model is missing
df_filtered = df_filtered.dropna(subset=['Model']).copy()

# Remove extra spaces and standardize model labels
df_filtered['Model'] = df_filtered['Model'].astype(str).str.strip()

# Remove rows where Model is empty
df_filtered = df_filtered[df_filtered['Model'] != ''].copy()

# Check the number of remaining records and missing models
print("Number of records after cleaning:", len(df_filtered))
print("Missing Model values:", df_filtered['Model'].isna().sum())

# Inspect the most common aircraft models
print("\nMost common aircraft models:")
print(df_filtered['Model'].value_counts().head(20))
```

```python

# Check whether each model belongs to more than one make

model_make_counts = df_filtered.groupby('Model')['Make'].nunique()

# Find models associated with multiple aircraft makes
shared_models = model_make_counts[model_make_counts > 1]

print("Number of models associated with multiple makes:", len(shared_models))
print("\nExamples of models associated with multiple makes:")
print(shared_models.head(20))
```

```python

# Create a unique identifier using the aircraft make and model
df_filtered['Aircraft.Type'] = (
    df_filtered['Make'].astype(str).str.strip().str.upper()
    + ' - ' +
    df_filtered['Model'].astype(str).str.strip().str.upper()
)

# Check the new column
print("Number of unique aircraft types:", df_filtered['Aircraft.Type'].nunique())
print("\nExamples of aircraft types:")
print(df_filtered['Aircraft.Type'].value_counts().head(20))
```

```python

# Inspect other columns for potential cleaning tasks

columns_to_inspect = [
    'Engine.Type',
    'Weather.Condition',
    'Number.of.Engines',
    'Purpose.of.flight',
    'Broad.phase.of.flight'
]

for column in columns_to_inspect:
    print(f"\n--- {column} ---")
    print("Missing values:", df_filtered[column].isna().sum())
    print("Distinct values:", df_filtered[column].nunique())
    print(df_filtered[column].value_counts(dropna=False).head(15))
```

```python

# Clean other columns
# Assumption: standardize text labels but preserve missing values
# because missing information does not mean an accident record is invalid.

# Standardize text columns by removing extra spaces
text_columns = [
    'Engine.Type',
    'Weather.Condition',
    'Purpose.of.flight',
    'Broad.phase.of.flight'
]

for column in text_columns:
    df_filtered[column] = df_filtered[column].apply(
        lambda value: value.strip() if isinstance(value, str) else value
    )

# Standardize known unknown labels in Engine.Type
df_filtered['Engine.Type'] = df_filtered['Engine.Type'].replace({
    'UNK': 'Unknown'
})

# Inspect the cleaned engine types
print("Engine types after cleaning:")
print(df_filtered['Engine.Type'].value_counts(dropna=False))

print("\nDataset shape after cleaning:", df_filtered.shape)
```
